```python
balance = 0

user = {
    'name': "Shahzada Waleed",
    'Age': 20,
}
def deposit():
    global balance
    print("\n===== Deposit An Amount =====")
    amount = int(input("Enter an Amount To Deposit: "))
    if amount <= 0:
        print("Please enter a positive amount.")
    else:
        balance = balance + amount
        print("Amount Deposited Successfully!")

def withdraw():
    global balance
    print("\n===== Withdrawing From An Account =====")
    amount = int(input("Enter an amount You Want To Withdraw: "))
    if amount <= 0:
        print("Please enter a positive amount.")
    elif amount > balance:
        print("Balance Is Insufficient...")
    else:
        balance = balance - amount
        print("Amount Withdrawn Successfully!")


def check_balance():
    print("\nYour Account Balance Is:", balance)

def account():
    print("\n===== Your Account Detail =====")
    print("Name:", user['name'])
    print("Age:", user['Age'])
    print("Account Balance:", balance)


while True:
    print("\n1. Deposit")
    print("2. Withdraw")
    print("3. Check Balance")
    print("4. Account")
    print("5. Exit")
    num = int(input("What do you want to do: "))
    if num == 1:
        deposit()
    elif num == 2:
        withdraw()
    elif num == 3:
        check_balance()
    elif num == 4:
        account()
    elif num == 5:
        print("\nThank you for using our banking system!")
        break
    else:
        print("\nInvalid option. Please try again.")
```
